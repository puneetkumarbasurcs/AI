```python
def dfs(start, goal):
    stack = [(start, [start])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            return path

        if state in visited:
            continue

        visited.add(state)

        z = state.index(0)
        r, c = divmod(z, 3)

        moves = []

        if r > 0:
            moves.append(z - 3)      
        if r < 2:
            moves.append(z + 3)      
        if c > 0:
            moves.append(z - 1)      
        if c < 2:
            moves.append(z + 1)      

        for nz in moves:
            new = list(state)
            new[z], new[nz] = new[nz], new[z]
            new = tuple(new)

            if new not in visited:
                stack.append((new, path + [new]))

    return None



def get_input():
    puzzle = []

    for i in range(3):
        row = list(map(int, input().split()))
        puzzle += row

    return tuple(puzzle)


print("Enter Initial State:")
start = get_input()

print("\nEnter Goal State:")
goal = get_input()



path = dfs(start, goal)



if path:
    print("\nSolution found!")
    print("Moves:", len(path) - 1)

    for i, state in enumerate(path):
        print("\nStep", i, ":")

        for j in range(0, 9, 3):
            print(state[j], state[j+1], state[j+2])

else:
    print("\nNo solution found")
```
