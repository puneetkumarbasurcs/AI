```python
def conflicts(board):
    count = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                count += 1

    return count


def show_board(board):
    n = len(board)

    for row in range(n):
        for col in range(n):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


def hill_climbing(n):
    board = [0, 0, 0, 0] if n == 4 else list(range(n))
    step = 1

    print("\nInitial Board:")
    show_board(board)
    print("Attacking pairs:", conflicts(board))

    while conflicts(board) != 0:
        current_cost = conflicts(board)
        best_board = board[:]
        best_cost = current_cost

        print("\nStep", step)
        print("Current cost:", current_cost)
        print("\nNeighbor costs:")

        for col in range(n):
            for row in range(n):
                if row != board[col]:
                    temp = board[:]
                    temp[col] = row
                    cost = conflicts(temp)

                    print("Move queen in column", col + 1,
                          "to row", row + 1,
                          "->", cost, "conflicts")

                    if cost < best_cost:
                        best_cost = cost
                        best_board = temp

        if best_cost == current_cost:
            print("No better neighbor found. Stopping.")
            print("Hill climbing is stuck at a local minimum.")
            return

        board = best_board
        print("\nBest neighbor selected:")
        show_board(board)
        print("Attacking pairs:", conflicts(board))

        step += 1

    print("\nSolution found!")
    show_board(board)
    print("Total attacking pairs:", conflicts(board))


n = int(input("Enter number of queens: "))

if n < 1:
    print("Invalid input")
else:
    hill_climbing(n)
```
