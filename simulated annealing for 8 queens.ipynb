```python

import random
import math

n = 8
board = []

# Get queen positions from the user
for i in range(n):
    row = int(input(f"Enter row for queen in column {i+1} (1-8): "))

    while row < 1 or row > 8:
        row = int(input("Enter row between 1 and 8: "))

    board.append(row - 1)

# Get initial temperature
T = float(input("Enter initial temperature (greater than 0): "))

while T <= 0:
    T = float(input("Enter temperature greater than 0: "))

# Get cooling factor
alpha = float(input("Enter cooling factor (between 0 and 1): "))

while alpha <= 0 or alpha >= 1:
    alpha = float(input("Invalid! Enter cooling factor between 0 and 1: "))

# Count attacking queen pairs
def conflicts(board):
    count = 0

    for i in range(n):
        for j in range(i + 1, n):
            if (board[i] == board[j] or
                abs(board[i] - board[j]) == abs(i - j)):
                count += 1

    return count

# Simulated Annealing
attempts = 0
max_attempts = 100000

while conflicts(board) > 0 and attempts < max_attempts and T > 0.01:
    new_board = board.copy()

    col = random.randint(0, n - 1)
    new_board[col] = random.randint(0, n - 1)

    delta = conflicts(new_board) - conflicts(board)

    if delta <= 0 or random.random() < math.exp(-delta / T):
        board = new_board

    T *= alpha
    attempts += 1

# Display final board
print("\nFinal board:")

for row in range(n):
    print(" ".join("Q" if board[col] == row else "." for col in range(n)))

print("Conflicts:", conflicts(board))

if conflicts(board) == 0:
    print("Solution found!")
else:
    print("Solution not found. Run again.")
```
